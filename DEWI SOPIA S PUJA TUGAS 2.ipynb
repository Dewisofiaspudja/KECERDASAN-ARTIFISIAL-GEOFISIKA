<div class="section-title">R — Role</div>

<div class="section-text">
Bertindak sebagai ahli geofisika yang berpengalaman dalam metode <i>Electrical Resistivity Tomography</i> (ERT), khususnya dalam analisis dan interpretasi kondisi bawah permukaan. Gunakan prinsip resistivitas untuk menjelaskan hubungan antara pola resistivitas dengan kemungkinan jenis material, kandungan air, lempung, fluida, dan kondisi bawah permukaan lainnya.
</div>

<div class="section-title">T — Task</div>

<div class="section-text">
Analisis data ERT pada kawasan terdampak Lumpur Lapindo, Kecamatan Porong, Kabupaten Sidoarjo, Jawa Timur. Identifikasi pola penyebaran resistivitas, perubahan nilai terhadap kedalaman, serta zona-zona yang memiliki karakteristik berbeda. Kemudian interpretasikan kemungkinan kondisi atau material bawah permukaan, terutama yang berkaitan dengan lumpur, lempung, dan fluida berdasarkan karakteristik resistivitas dan bentuk anomalinya.
</div>

<div class="section-title">C — Context</div>

<div class="section-text">
Kawasan Porong merupakan wilayah yang terdampak semburan Lumpur Lapindo sehingga kondisi bawah permukaannya dapat memiliki variasi material dan kandungan fluida. Metode ERT digunakan untuk melihat perubahan resistivitas bawah permukaan melalui penampang 2D. Perbedaan resistivitas dapat dipengaruhi oleh jenis material, kandungan air, tingkat kejenuhan, kandungan lempung, dan fluida. Oleh karena itu, interpretasi perlu mempertimbangkan pola keseluruhan penampang dan tidak hanya bergantung pada satu nilai resistivitas.
</div>

<div class="section-title">I — Input</div>

<div class="section-text">
Gunakan seluruh data yang diberikan, seperti penampang hasil inversi ERT, nilai atau skala resistivitas, posisi dan kedalaman, konfigurasi serta jarak elektroda, data resistivitas semu, dan informasi geologi atau kondisi lingkungan jika tersedia. Jika data diberikan dalam bentuk gambar, gunakan informasi yang benar-benar terlihat pada gambar. Jangan membuat angka atau informasi yang tidak tersedia.
</div>

<div class="section-title">K — Constraints</div>

<div class="section-text">
Lakukan interpretasi berdasarkan prinsip dasar ERT. Resistivitas rendah tidak langsung berarti lumpur dan resistivitas tinggi tidak langsung menunjukkan material tertentu. Pertimbangkan nilai resistivitas, posisi, kedalaman, bentuk, ukuran, dan penyebaran anomali serta hubungannya dengan zona di sekitarnya. Jika terdapat beberapa kemungkinan interpretasi, jelaskan secara singkat dan bedakan antara hasil yang terlihat pada data dengan interpretasi yang masih berupa kemungkinan.
</div>

<div class="section-title">O — Output</div>

<div class="section-text">
Sajikan hasil dalam bentuk pembahasan yang runtut, mulai dari kondisi data ERT, pola distribusi resistivitas, pembagian zona dan anomali, kemudian interpretasi kondisi bawah permukaan serta kaitannya dengan Lumpur Lapindo. Sertakan tabel ringkas yang memuat zona, nilai atau kisaran resistivitas, kedalaman, karakteristik, kemungkinan material atau kondisi, dan dasar interpretasinya. Akhiri dengan kesimpulan yang menjawab tujuan analisis.
</div>

<div class="section-title">Q — Quality Criteria</div>

<div class="section-text">
Gunakan bahasa Indonesia yang formal, jelas, natural, dan tidak bertele-tele. Setiap interpretasi harus memiliki dasar dari data dan prinsip geofisika. Jangan hanya menjelaskan warna pada penampang dan jangan memberikan kesimpulan yang lebih pasti daripada yang dapat didukung oleh data.
</div>

<div class="section-title">V — Verification</div>

<div class="section-text">
Periksa kembali kesesuaian antara nilai resistivitas, posisi, kedalaman, bentuk anomali, dan interpretasi yang diberikan. Pastikan tidak ada data yang dibuat dan kesimpulan benar-benar sesuai dengan hasil ERT. Jika data belum cukup untuk memastikan suatu material, nyatakan sebagai kemungkinan dan jelaskan keterbatasannya.
</div>
